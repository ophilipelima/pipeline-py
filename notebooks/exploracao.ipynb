```python
path_json = '../data_raw/dados_empresaA.json' 
```

```python
import json
```

```python
with open(path_json, 'r') as file:
    dados_json = json.load(file) #lista de dict
dados_json[1]
```

```python
path_csv = '../data_raw/dados_empresaB.csv'
```

```python
import csv 

dados_csv = []

with open(path_csv, 'r') as file:
    spamreader = csv.DictReader(file, delimiter=',')
    for linha in spamreader:
      dados_csv.append(linha)
dados_csv[0]
```

```python
nomes_colunas_json = list(dados_json[0].keys())
nomes_colunas_csv = list(dados_csv[0].keys())

print(nomes_colunas_json)
print(nomes_colunas_csv)
```

```python
key_mapping = {'Nome do Item': 'Nome do Produto',
                'Classificação do Produto': 'Categoria do Produto',
                'Valor em Reais (R$)': 'Preço do Produto (R$)',
                'Quantidade em Estoque': 'Quantidade em Estoque',
                'Nome da Loja': 'Filial',
                'Data da Venda': 'Data da Venda'}
```

```python
novos_dados_csv = [{key_mapping.get(chaveAntiga): valor for chaveAntiga, valor in dictAntigo.items()} for dictAntigo in dados_csv]

print(novos_dados_csv[0])

#novos_dados_csv = [{key_mapping.get(chaveAntiga, chaveAntiga): valor for chaveAntiga, valor in dictAntigo.items()} for dictAntigo in dados_csv]
        
```

```python
lista_combinada = []

lista_combinada.extend(dados_json)
lista_combinada.extend(novos_dados_csv)
print(len(lista_combinada))
```

```python

print(list(lista_combinada[-1].keys()))
```

```python
nomes_colunas = list(lista_combinada[-1].keys())

print(nomes_colunas)
```

```python
dados_combinados_tabela = [nomes_colunas]

for row in lista_combinada:
    linha = []
    for coluna in nomes_colunas:
        linha.append(row.get(coluna, 'Indisponível'))
    dados_combinados_tabela.append(linha)
        
print(dados_combinados_tabela[0])
print(dados_combinados_tabela[1])
print(dados_combinados_tabela[-1])
```

```python
path_dados_combinados = '../data_processed/dados_combinados.csv'

with open(path_dados_combinados, 'w') as file: 
    escritor = csv.writer(file)
    escritor.writerows(dados_combinados_tabela)
```
